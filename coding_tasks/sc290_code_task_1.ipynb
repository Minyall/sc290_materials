```python
#*
# DO NOT EDIT THIS CELL. RUN IT AND CONTINUE READING THE INSTRUCTIONS BELOW

# Necessary imports for the function
import requests
from scipy.spatial.distance import pdist, squareform
from scipy.sparse._csr import csr_matrix
import plotly.express as px
import numpy as np
import pandas as pd

# Helper function
def check_my_vectors(vectors):
    if type(vectors) is pd.DataFrame:
        print('You do not need to wrap your vectors in a Pandas DataFrame')
        return
    if type(vectors) is csr_matrix:
        vectors = vectors.todense()
    if type(vectors) is np.ndarray:
        labels = [f'Pet_{x}' for x in range(5)] + [f'Sub_{x}' for x in range(5)]
        distances = squareform(pdist(vectors, metric='cosine'))
        px.imshow(pd.DataFrame(distances, columns =labels, index=labels), color_continuous_scale='BuGn_r', 
                title='Heatmap of document similarity<br><sub>0=Similar|1=Different<br>').show()
    else:
        print('Your vectors variable has not been created using a technique taught in the course')

# Loading the task data
response = requests.get('https://github.com/Minyall/sc207_290_public/raw/refs/heads/main/data/pets_vs_submarines.txt')
texts = [x.strip() for x in response.text.split('\n\n')]
```
